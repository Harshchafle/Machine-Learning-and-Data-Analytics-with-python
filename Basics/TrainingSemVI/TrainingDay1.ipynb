```python
# Non-default argument follows default argument
def operate(a, b, op="+"):
  if op == "+":
    return a+b
  elif op == "-":
    return abs(a-b)
  elif op == "*":
    return a*b
  elif op == "/":
    return a/b if b!=0 else "den is zero"

print(operate(10, 20, "/"))
```

```python
def argwlafunc(*args):
  sum = 0
  for num in args:
    sum += num
  print("sum in args : ", sum)
  print(args)

print(argwlafunc(1,2,3,4,5,6,7))
```

```python
def kwargwlafunc(**kwargs):
  sum = 0
  for key, value in kwargs.items():
    sum += value
  print(f"sum of values : {sum}")
  return kwargs, type(kwargs)

kwargwlafunc(a=1,b=2)
```

```python
# generator -> yeild use krne ke baad hi koi bhi funct generator bnta hai or execute bhi hona chahiyea
def gen():
  yield(5)
  yield(10)
g = gen()
print(g, type(g))
print(next(gen()))
print(next(gen()))
print(next(g))
print(next(g))
```

```python
def gen():
  for i in range(0, 10, 2):
    yield(i)
g = gen()
print(g, type(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
#print(next(g)) -> err = StopIterator
```

```python
def myrange(st, end, step):
  i = st
  while False:
    yield(i)
    i += step

g = myrange(0, 10, 2)
print(g, type(g))
print(type(myrange(0,10,2)))
```

# custom Iterator

```python
class MyIterator:   # generate (count) number of random values between low and high
  def __init__(self, low, high, count):
    self.low = low
    self.high = high
    self.count = count
    self.counter = 0

  def __iter__(self):
    return self

  def __next__(self):
    if self.counter < self.count:
      self.counter += 1
      return self.low + random() * (self.high - self.low)
    else:
      raise StopIteration

```

```python
for n in MyIterator(10, 25, 5):
  print(n)
```
