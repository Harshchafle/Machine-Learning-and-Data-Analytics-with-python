```python
# 🔥 Monkey patch to fix wordcloud + matplotlib crash
import numpy as np

_original_asarray = np.asarray

def _safe_asarray(a, *args, **kwargs):
    kwargs.pop("copy", None)  # REMOVE the bad argument
    return _original_asarray(a, *args, **kwargs)

np.asarray = _safe_asarray
```
