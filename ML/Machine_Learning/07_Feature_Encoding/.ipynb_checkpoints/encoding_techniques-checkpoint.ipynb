```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('customer.csv')
df.sample(5)
```

```python
df = df.iloc[:, 2:]
df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2], df.iloc[:,-1], test_size=0.2, random_state=0)
```

# ordinal Encoder

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Poor', 'Average', 'Good'],['School', 'UG', 'PG']])
```

```python
oe.fit(X_train)
```

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
(X_train[0], X_train[1], X_train[2])
```

```python
oe.categories_
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder # le is only for target class not for inputs
le = LabelEncoder()

le.fit(y_train)
```

```python
le.classes_
```

```python
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```

## One Hot Encoding using Sklearn

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:4], df.iloc[:, -1], test_size=0.2, random_state=0)
X_train.sample(5)
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first')
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel', 'owner']]).toarray()
```

```python
X_test_new = ohe.transform(X_test[['fuel', 'owner']]).toarray()
```
