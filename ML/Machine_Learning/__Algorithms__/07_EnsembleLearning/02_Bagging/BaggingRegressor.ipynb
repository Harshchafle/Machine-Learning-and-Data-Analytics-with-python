```python
from sklearn import datasets

wine = datasets.load_wine()
X_wine, y_wine = wine.data, wine.target
print('Features : ', str(wine.feature_names))
print("feature_size : ", str(wine.data.shape))
print('target size : ', str(wine.target.size))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_wine, y_wine, test_size=0.2, random_state=123)
print('Train/Test Sets Sizes : ', X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
knn = KNeighborsRegressor()
```

```python
lr.fit(X_train, y_train)
```

```python
dt.fit(X_train, y_train)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred_lr = lr.predict(X_test)
y_pred_dt = dt.predict(X_test)
y_pred_knn = knn.predict(X_test)
```

```python
print('R2_score of LR : ', r2_score(y_test, y_pred_lr))
print('R2_score of DT : ', r2_score(y_test, y_pred_dt))
print('R2_score of KNN : ', r2_score(y_test, y_pred_knn))
```

```python
from sklearn.ensemble import BaggingRegressor

bag_regressor = BaggingRegressor(random_state=1)
bag_regressor.fit(X_train, y_train)
```

```python
y_pred = bag_regressor.predict(X_test)

print('Training Coefficient of R2 : %.3f'%bag_regressor.score(X_train, y_train))
print('Test Coefficient of R2 : %.3f'%bag_regressor.score(X_test, y_test))
```

```python
print('Train R2 score : %.3f'%bagging_regressor_grid.best_estimator_.score(X_train, y_train))
print('Test R2 score : %.3f'%bagging_regressor_grid.best_estimator_.score(X_test, y_test))
print('Best R2 score through Grid Search : %.3f'%bagging_regressor_grid.best_score_)
print('Best Parameters : ', bagging_regressor_grid.best_params_)
```
