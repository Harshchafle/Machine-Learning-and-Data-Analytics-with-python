```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
```

### 1. Perceptron Trick

```python
def perceptron(X, y):
  X = np.insert(X, 0, 1, axis=1)
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(1000):
    j = np.random.randint(0, 100)
    y_hat = step(np.dot(X[j], weights))
    weights = weights + lr*(y[j] - y_hat)*X[j]

  return weights[0], weights[1:]
```

```python
def step(z):
  return 1 if z > 0 else 0
```

```python
intercept_, coef_ = perceptron(X,y)
```

```python
print(coef_, intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3, 3, 100)
y_input1 = m*x_input1 + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input1, y_input1, color='red', linewidth=3)
plt.scatter(X[:,0],X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3,2)
```

### 2. sklearn Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = -(lor.intercept_/lor.coef_[0][1])
```

```python
x_input2 = np.linspace(-3, 3, 100)
y_input2 = m*x_input2 + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input1, y_input1, color='red', linewidth=3)
plt.plot(x_input2, y_input2, color='black', linewidth=3)
plt.scatter(X[:,0],X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3,2)
```

### Updation in Perceptron Trick : Using Sigmoid activation instead of Step function as it gives non zero value for any input.

```python
def perceptron(X, y):
  X = np.insert(X, 0, 1, axis=1)
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(1000):
    j = np.random.randint(0, 100)
    y_hat = sigmoid(np.dot(X[j], weights))
    weights = weights + lr*(y[j]-y_hat)*X[j]

  return weights[0], weights[1:]
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
intercept_, coef_ = perceptron(X, y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input3 = np.linspace(-3, 3, 100)
y_input3 = m*x_input3 + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input1, y_input1, color='red', linewidth=3)
plt.plot(x_input2, y_input2, color='yellow', linewidth=3)
plt.plot(x_input3, y_input3, color='orange', linewidth=3)
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3, 2)
```

### Comparing the sklearn LogisticRegressioin and applying Gradient Descent

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression(penalty=None, solver='sag')
lor.fit(X,y)
```

```python
print(lor.coef_)
print(lor.intercept_)
```

```python
m1 = -(lor.coef_[0][0]/lor.coef_[0][1])
b1 = -(lor.intercept_/lor.coef_[0][1])
```

```python
x_input = (np.linspace(-3, 3, 100))
y_input = m1*x_input + b1
```

```python
def GD(X,y):
  X = np.insert(X, 0, 1, axis = 1)
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(2500):
    y_hat = sigmoid(np.dot(X,weights))
    weights = weights + lr * (np.dot((y-y_hat), X) / X.shape[0])

  return weights[1:], weights[0]
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
coef_, intercept_ = GD(X, y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3, 3, 100)
y_input1 = m*x_input1 + b
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color='black', linewidth=3) # lOR(sklearn)
plt.plot(x_input1, y_input1, color='red', linewidth=3) # GD
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3,2)
```

## Performance metrics and Confusion Matrix


```python
from sklearn.metrics import accuracy_score, r2_score, f1_score, precision_recall_fscore_support, confusion_matrix
```

```python
y_pred = lor.predict(X)
print('Accuracy : ',accuracy_score(y, y_pred))
print('R2 score : ',r2_score(y, y_pred))
print('f1 score : ',f1_score(y, y_pred))
print('Precision : ',precision_recall_fscore_support(y, y_pred))
print('CF : \n', confusion_matrix(y, y_pred))
print('Here the accuracy and all will be 1 because we have trained data and predicting on it again')
```
