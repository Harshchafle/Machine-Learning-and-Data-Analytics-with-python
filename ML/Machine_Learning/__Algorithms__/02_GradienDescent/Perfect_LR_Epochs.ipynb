```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X, y)

print(lr.coef_)
print(lr.intercept_)
```

```python
m = 59.67
```

```python
# defining class of GD assuming m is constant
class GDRegressor:
    def __init__(self, lr, epochs):
        self.lr = lr
        self.epochs = epochs
        self.b = -120
        self.m = m

    def fit(self, X, y):
        for i in range (self.epochs):
            loss_slope = -2*np.sum((y - m*X.ravel() - self.b))
            self.b = self.b - (self.lr * loss_slope)
            print(self.b)
        print("fitting done!!!")
```

```python
model = GDRegressor(0.1, 10)
model.fit(X,y)
```

```python
model = GDRegressor(0.01, 10)
model.fit(X,y)
```

```python
model = GDRegressor(0.001, 10)
model.fit(X, y)
```

```python
model = GDRegressor(0.001, 100)
model.fit(X,y)
```
