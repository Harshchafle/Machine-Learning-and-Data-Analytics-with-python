```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
print(X.shape, y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
y_pred = model.predict(X_test)
print(r2_score(y_test, y_pred))
```

```python
class SGDRegressor:
    def __init__(self, lr=0.01, epochs=100):
        self.lr = lr
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        # init coef_ and intercept_
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0

        for i in range(self.epochs):
            idx = np.random.randint(0, X_train.shape[0])
            
            y_hat = np.dot(X_train[idx], self.coef_) + self.intercept_
            
            intercept_der = -2*(y_train[idx] - y_hat)
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)

            coef_der = -2*np.dot((y_train[idx] - y_hat), X_train[idx])
            self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_, self.coef_)

    
    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
sgdr = SGDRegressor(0.1, 10000)
```

```python
sgdr.fit(X_train, y_train)
```

```python
y_pred = sgdr.predict(X_test)
print(r2_score(y_test, y_pred))
```

```python
# SGD is faster than BATCH GD as it takes less epoches but if epoches are same batch GD is faster
```

## Learning Schedule : where we vary the lr with epochs

```python
t0, t1 = 5, 50
epochs = 50
def learning_rate(t):
    return t0/(t + t1)

for i in range(epochs):
    for j in range(X.shape[0]):
        lr = learning_rate(i * X.shape[0] + j) 
print(lr)
```
