```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X, y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y)
```

## Applying Ordinary Least Square (OLS)

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X, y)

LinearRegression(copy_X=True, fit_intercept=True, n_jobs=None)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red')
```

## Applying Gradient Descent


```python
# GD by assuming m = 78.35
# assume starting value of b = 0
y_pred = ((model.coef_*X + 0).reshape(4))
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red', label='OLS')
plt.plot(X, y_pred, color='#00a52a', label='b = 0')
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_slope = -2*np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
# taking learning_rate = 0.1
# n.slope --> step size
lr = 0.1
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((model.coef_*X + b).reshape(4))

plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red', label='OLS')
plt.plot(X, y_pred, color='#00a52a', label='b = 0')
plt.plot(X, y_pred1, color='#45aa90', label='b = {}'.format(b))
plt.legend()
plt.show()
```

```python
# iteration 2
loss_slope = -2*np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((model.coef_*X + b).reshape(4))

plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red', label='OLS')
plt.plot(X, y_pred, color='#00a52a', label='b = 0')
plt.plot(X, y_pred1, color='#45aa90', label='b = 20.927763408209216')
plt.plot(X, y_pred2, color='#000000', label='b = {}'.format(b))

plt.legend()
plt.show()
```

```python
# iteration 3
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((model.coef_*X + b).reshape(4))

plt.scatter(X,y)
plt.plot(X, model.predict(X), color='red', label='OLS')
plt.plot(X, y_pred, color='#00a52a', label='b = 0')
plt.plot(X, y_pred1, color='#45aa90', label='b = 20.927763408209216')
plt.plot(X, y_pred2, color='#00aa90', label='b = 25.11331608985106')
plt.plot(X, y_pred3, color='#8888bb', label='b = {}'.format(b))

plt.legend()
plt.show()
```

```python
b = -200
m = 78.35
lr = 0.1

epochs = 15
plt.scatter(X,y)
plt.plot(X,model.predict(X), color='red', label='OLS')
for epoch in range(epochs):
    loss_slope = -2*np.sum((y - m*X.ravel() - b).reshape(4))
    step_size = loss_slope*lr
    b = b - step_size
    y_pred = ((model.coef_*X.ravel() + b).reshape(4))
    plt.plot(X, y_pred, label='b = {}'.format(b))

plt.show()
```

```python
# Remember steps
# 1. Calculate loss_slope = -2*np.sum((y - m*X.ravel() - b).reshape(4))
# 2. Step_size = loss_slope * lr
# 3. update b = b - step_size    --> bnew = bold - lr*loss_slope
# 4. predict on new b y_pred = ((model.coef_*X.ravel() + b).reshape(4))
```
