```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X, y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20, random_state=13)

X_train,X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=13)

plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X_train, y_train)
print(reg.coef_)
print(reg.intercept_)
```

```python
alphas = [0, 1, 5, 10, 50, 100]
plt.figure(figsize=(6, 4))
plt.scatter(X, y)

for i in alphas:
  L = Lasso(alpha=i)
  L.fit(X_train, y_train)
  plt.plot(X_test, L.predict(X_test), label='alpha = {}'.format(i))
plt.legend()
plt.show()
```

```python
m =100
x1 = 5*np.random.rand(m, 1) - 2
x2 = 0.7*x1**2 - 2*x1+3+np.random.randn(m,1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

def get_preds_lasso(x1, x2, alpha):
  model = Pipeline([
      ('poly_feats', PolynomialFeatures(degree=16)),
      ('lasso', Lasso(alpha=alpha))
  ])

  model.fit(x1, x2)
  return model.predict(x1)

alphas = [0, 0.1, 1]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10,6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
  preds = get_preds_lasso(x1, x2, alpha)
  plt.plot(sorted(x1[:,0]), preds[np.argsort(x1[:, 0])], c, label='Alpha : {}'.format(alpha))

plt.legend()
plt.show()
```

### Lasso Regression key POints

### 1. How are coefficients affected?

```python
from sklearn.datasets import load_diabetes

from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score
```

```python
data = load_diabetes()

df = pd.DataFrame(data.data, columns=data.feature_names)
df['TARGET'] = data.target
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=2)
```

```python
coefs = []
r2_scores = []

for i in  [0, 0.1, 1, 2, 3, 5]:
  reg = Lasso(alpha=i)
  reg.fit(X_train, y_train)
  coefs.append(reg.coef_.tolist())
  y_pred = reg.predict(X_test)
  r2_scores.append(r2_score(y_test, y_pred))
```

```python
print(coefs)
```

```python
plt.figure(figsize=(14, 9))
plt.subplot(231)
plt.bar(data.feature_names, coefs[0])
plt.title('Alpha = 0, r2 = {}'.format(round(r2_scores[0], 2)))

plt.subplot(232)
plt.bar(data.feature_names, coefs[1])
plt.title('Alpha = 0.1, r2 = {}'.format(round(r2_scores[1], 2)))

plt.subplot(233)
plt.bar(data.feature_names, coefs[2])
plt.title('Alpha = 1, r2 = {}'.format(round(r2_scores[2], 2)))

plt.subplot(234)
plt.bar(data.feature_names, coefs[3])
plt.title('Alpha = 2, r2 = {}'.format(round(r2_scores[3], 2)))

plt.subplot(235)
plt.bar(data.feature_names, coefs[4])
plt.title('Alpha = 3, r2 = {}'.format(round(r2_scores[4], 2)))

plt.subplot(236)
plt.bar(data.feature_names, coefs[5])
plt.title('Alpha = 5, r2 = {}'.format(round(r2_scores[5], 2)))
```

### 4. Effect of regulariation on loss function

```python
from sklearn.datasets import make_regression

X, y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=25, random_state=13)

plt.scatter(X, y)

from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X, y)
print(reg.coef_)
print(reg.intercept_)
```

```python
def cal_loss(m, alpha):
    return np.sum((y-m*X.ravel() + 2.29)**2) + alpha*abs(m)
```

```python
def predict(m):
    return m*X - 2.29
```

```python
m = np.linspace(-45, 100, 100)
plt.figure(figsize=(10, 6))
for j in [0, 100, 500, 1000, 1500, 2000, 2500, 4000, 5000]:
    loss = []
    for i in range(m.shape[0]):
        loss_i = cal_loss(m[i], j)
        loss.append(loss_i)
    plt.plot(m, loss, label='alpha = {}'.format(j))
plt.legend()
plt.xlabel('m')
plt.ylabel('Loss')
plt.show()
```
