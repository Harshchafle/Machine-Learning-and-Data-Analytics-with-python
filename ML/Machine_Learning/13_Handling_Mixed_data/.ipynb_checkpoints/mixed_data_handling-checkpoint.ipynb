```python
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('../__datasets__/train.csv', usecols=['Survived','Cabin', 'Ticket'])
```

```python
import random
df['numbers'] = 2
for i in range(len(df)):
    num = random.randint(0,10)
    df['numbers'][i] = num if num != 1 else 'A'
df.head()
```

```python
df['numbers'].unique()
```

```python
fig = df['numbers'].value_counts().plot.bar()
fig.set_title('Passengers travelling with')
```

```python
# extract numerical part 
df['num_num'] = pd.to_numeric(df['numbers'], errors='coerce', downcast='integer')
# 'A' -> NAN
# nums -> nums
```

```python
df['num_cat'] = np.where(df['num_num'].isnull(), df['numbers'], np.nan)
```

```python
df.sample(10)
```

```python
df['Cabin'].unique()
```

```python
df['Cabin_num'] = df['Cabin'].str.extract('(\d+)')
df['Cabin_cat'] = df['Cabin'].str[0]
df.sample(10)
```

```python
df['Cabin_cat'].value_counts().plot(kind='bar')
```

```python
df['Ticket'].unique()
```

```python
df['Ticket'].nunique()
```

```python
df['Ticket_num'] = df['Ticket'].apply(lambda s : s.split()[-1])
df['Ticket_num'] = pd.to_numeric(df['Ticket_num'], errors='coerce', downcast='integer')

df['Ticket_cat'] = df['Ticket'].apply(lambda s : s.split()[0])
df['Ticket_cat'] = np.where(df['Ticket_cat'].str.isdigit(), np.nan, df['Ticket_cat'])
```

```python
df.sample(10)
```

```python
df['Ticket_cat'].value_counts()
```
