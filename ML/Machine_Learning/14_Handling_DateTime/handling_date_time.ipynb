## Working with dates

```python
# converting to datetime datatypes
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

## Extracting Year

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

## Extacting Month

```python
date['date_month'] = date['date'].dt.month

date.sample(5)
```

## Extracting Month by name

```python
date['Month'] = date['date'].dt.month_name()
date.sample(5)
```

## Extracting Days

```python
date['day'] = date['date'].dt.day
date.sample(5)
```

## Extracting which day of week

```python
date['dayno'] = date['date'].dt.dayofweek
date.sample(5)
```

## Extracting Day Name

```python
date['DayName'] = date['date'].dt.day_name()
date.sample(5)
```

## Extracting weekends

```python
date['weekends'] = np.where(date['DayName'].isin(['Sunday', 'Saturday']), 1, 0)
date.sample(3)
```

## Extract week of the year

```python
date['weekNo'] = date['date'].dt.isocalendar().week
date.sample(3)
```

## Extract Quarter

```python
date['Quarter'] = date['date'].dt.quarter
date.sample(3)
```

## Extracting Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
inf = (today - date['date'])
```

```python
inf.dt.days
```

```python
# Monthd Passed
np.round((inf) / np.timedelta64(1, 'M'), 0)
```
