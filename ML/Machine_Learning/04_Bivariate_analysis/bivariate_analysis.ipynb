```python
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings
```

```python
tips = sns.load_dataset('tips')
```

```python
titanic = pd.read_csv('train.csv')
```

```python
flights = sns.load_dataset('flights')
```

```python
iris = pd.read_csv('Iris.csv')
```

## 1. ScatterPlot (NUmerical - Numerical)

```python
tips.sample()
```

```python
sns.scatterplot(data = tips, x='total_bill', y='tip', hue=tips['sex'], style=tips['smoker'], size = tips['size'
    ])
# hue adds additional information abouts data points, how that data point is associated with the column passed in hue
# style shows the different values of columns passed in style
# size adjusts the size of data point according to the column passed in it
```

## 2. BarPlot (Numerical - Categorical)

```python
titanic.sample()
```

```python
sns.barplot(data = titanic, x='Pclass', y='Fare', hue=titanic['Sex'])
```

## 3. BoxPlot (Numerical - Categorical)

```python
sns.boxplot(data = titanic, x='Age', y='Sex', hue=titanic['Pclass'])
```

## 4. DistPlot (Numerical - Categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'])
```

```python
sns.distplot(titanic[titanic['Survived']==1]['Age'])
```

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
```

## 5. HeatMap (Categorical - Categorical)

```python
titanic.sample()
```

```python
matrix = pd.crosstab(titanic['Pclass'], titanic['Survived'])
sns.heatmap(matrix)
```

```python
(titanic.groupby('Pclass').mean(numeric_only=True)['Survived']*100).plot(kind='bar')
```

```python
titanic.groupby('Embarked').mean(numeric_only=True)
```

## 6. Clustermap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['SibSp'], titanic['Survived']))
# dendrogram representation
```

## 7. PairPlot

```python
iris.sample()
```

```python
sns.pairplot(data=iris, hue='Species')
```

## 8. LinePlot (Numerical - Numerical)

```python
flights.sample()
```

```python
data = flights.groupby('year', as_index=False)['passengers'].sum().reset_index()
# print(data)
sns.lineplot(data, x='year', y='passengers')
```

```python
matrix = flights.pivot_table(values='passengers', index='month', columns='year')
matrix
```

```python
sns.heatmap(matrix)
```

```python
 sns.clustermap(matrix)
```
